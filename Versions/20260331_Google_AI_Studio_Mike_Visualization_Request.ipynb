```python
import pandas as pd
import plotly.express as px
import numpy as np

# 1. GENERATE MULTI-SECTOR 10-YEAR HISTORICAL DATA (2014-2030)
# This simulates the Open Data DC "Economic Development" dataset
np.random.seed(42)
years = np.arange(2014, 2031)
sectors = ['Residential', 'Office', 'Retail', 'Hotel', 'Mixed-Use']
wards = [1, 2, 3, 4, 5, 6, 7, 8]

data_list = []
for _ in range(300): # 300 major projects across DC
    year = np.random.choice(years)
    sector = np.random.choice(sectors)
    ward = np.random.choice(wards)

    # Simulate DC Geographies (approximate lat/long for Wards)
    lat = 38.89 + np.random.uniform(-0.05, 0.05)
    lon = -77.01 + np.random.uniform(-0.05, 0.05)

    # Scale: Residential in units, Office/Retail in SqFt (Normalized for bubble size)
    scale = np.random.randint(50, 1200) if sector == 'Residential' else np.random.randint(10000, 500000) / 500

    data_list.append({
        "Project_Name": f"Project_{np.random.randint(1000, 9999)}",
        "Year": year,
        "Sector": sector,
        "Ward": f"Ward {ward}",
        "Lat": lat,
        "Lon": lon,
        "Size_Metric": scale,
        "Status": "Completed" if year <= 2024 else "Pipeline"
    })

df = pd.DataFrame(data_list)

# 2. THE "FLOOD" LOGIC: CUMULATIVE TIME SERIES
# To show the city "flooding" with growth, each frame must contain all previous years
frames = []
for y in sorted(df['Year'].unique()):
    temp_df = df[df['Year'] <= y].copy()
    temp_df['Display_Year'] = y
    frames.append(temp_df)

flood_df = pd.concat(frames)

# 3. ADVANCED GEO-INTELLIGENCE VISUALIZATION
fig = px.scatter_mapbox(
    flood_df,
    lat="Lat",
    lon="Lon",
    size="Size_Metric",
    color="Sector",      # MULTI-SECTOR ANALYSIS
    hover_name="Project_Name",
    hover_data=["Year", "Ward", "Status"],
    animation_frame="Display_Year", # 10-YEAR TIME SERIES
    title="DC Decade of Growth: Multi-Sector 'Flood Map' (2014-2030)",
    mapbox_style="carto-dark", # Dark mode makes the "flood" colors pop like the NYT example
    category_orders={"Display_Year": sorted(df['Year'].unique())},
    color_discrete_map={
        "Residential": "#00d2ff", # Blue
        "Office": "#ff4b2b",      # Red
        "Retail": "#f9d423",      # Yellow
        "Hotel": "#a445b2",       # Purple
        "Mixed-Use": "#2ecc71"    # Green
    },
    zoom=11,
    center={"lat": 38.895, "lon": -77.01},
    height=800
)

# Professional Styling
fig.update_layout(
    margin={"r":0,"t":50,"l":0,"b":0},
    legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01, bgcolor="rgba(0,0,0,0.5)", font_color="white")
)

fig.show()
```
